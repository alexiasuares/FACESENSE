```python
import pandas as pd
from collections import Counter
import numpy as np

# === 1. Carregar labels ===
labels_df = pd.read_csv('../../../data/iMIGUE/Label/labels_20200831.csv')

# Reconstruir o label_to_int e int_to_label
unique_labels = sorted(labels_df["class"].unique())
label_to_int = {label: idx for idx, label in enumerate(unique_labels)}
int_to_label = {v: k for k, v in label_to_int.items()}

# === 2. Carregar y_train, y_val, y_test ===
y_train = np.load('../../../data/processed_data/y_train.npy')
y_val   = np.load('../../../data/processed_data/y_val.npy')
y_test  = np.load('../../../data/processed_data/y_test.npy')

# === 3. Contar frequência em cada split ===
counter_train = Counter(y_train)
counter_val   = Counter(y_val)
counter_test  = Counter(y_test)

print("Distribuição y_train:", counter_train)
print("Distribuição y_val:", counter_val)
print("Distribuição y_test:", counter_test)

# === 4. Descobrir classes com menos de 5 amostras ===
classes_menos_5 = [cls for cls, count in counter_train.items() if count < 5]

print("\nClasses com menos de 5 amostras (no train):")
for c in classes_menos_5:
    print(f"Classe {c} → {int_to_label.get(c, 'Desconhecido')} (freq: {counter_train[c]})")
```
